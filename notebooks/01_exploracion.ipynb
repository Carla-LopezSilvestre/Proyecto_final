```python
import pandas as pd

ventas = pd.read_csv("../data/raw/sales.csv")
productos = pd.read_csv("../data/raw/products.csv")
tiendas = pd.read_csv("../data/raw/stores.csv")

print("ventas:", ventas.shape)
print("productos:", productos.shape)
print("tiendas:", tiendas.shape)
```

```python
print("--- VENTAS ---")
print(ventas.head())
print(ventas.dtypes)

print("\n--- PRODUCTOS ---")
print(productos.head())
print(productos.dtypes)

print("\n--- TIENDAS ---")
print(tiendas.head())
print(tiendas.dtypes)
```

```python
# 1) Fechas: de texto a fecha
ventas["Date"] = pd.to_datetime(ventas["Date"])
tiendas["Store_Open_Date"] = pd.to_datetime(tiendas["Store_Open_Date"])

# 2) Precios: quitar el "$" y los espacios, y pasar a número
for col in ["Product_Cost", "Product_Price"]:
    productos[col] = (
        productos[col]
        .str.replace("$", "", regex=False)
        .str.strip()
        .astype(float)
    )

# Comprobación
print(ventas.dtypes)
print(productos.dtypes)
print(tiendas.dtypes)
print(productos.head())
```

```python
tablas = {"ventas": ventas, "productos": productos, "tiendas": tiendas}

for nombre, df in tablas.items():
    print(f"--- {nombre.upper()} ---")
    print("Valores vacíos por columna:")
    print(df.isna().sum())
    print("Filas duplicadas:", df.duplicated().sum())
    print()

print("--- RESUMEN DE VENTAS ---")
print(ventas[["Units"]].describe())
print("Rango de fechas:", ventas["Date"].min(), "->", ventas["Date"].max())

print("\n--- CATEGORÍAS Y UBICACIONES ---")
print(productos["Product_Category"].value_counts())
print(tiendas["Store_Location"].value_counts())
```

```python
productos_raw = pd.read_csv("../data/raw/products.csv")
tiendas_raw = pd.read_csv("../data/raw/stores.csv")

with pd.ExcelWriter("../data/raw/maestro_tiendas_productos.xlsx") as writer:
    productos_raw.to_excel(writer, sheet_name="productos", index=False)
    tiendas_raw.to_excel(writer, sheet_name="tiendas", index=False)

# Comprobación: leerlo de vuelta
hojas = pd.read_excel("../data/raw/maestro_tiendas_productos.xlsx", sheet_name=None)
for nombre, df in hojas.items():
    print(nombre, df.shape)
    
```
